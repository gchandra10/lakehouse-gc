# 01 · Easy · Quick start

Four read-only steps: check what you can read, see the rules, scan, and look at the
results. An optional fifth step saves them, only to the table or file you name.

**Before you start:** upload the wheel and `policy.yaml` to a volume, then set the
paths in the next two cells.

```python
%pip install /Volumes/main/default/lakehouse_gc/lakehouse_gc-0.3.0-py3-none-any.whl
```

```python
from lakehouse_gc import check_access, rules_dataframe, run

POLICY_PATH = "/Volumes/main/default/lakehouse_gc/policy.yaml"

# Step 1: what can I read here? This probes column names only; it scans nothing.
access = check_access(spark, POLICY_PATH, scanners=["uc"])
print(access.summary())
```

```python
# Step 2: what will be checked, and with which settings from my policy?
display(rules_dataframe(spark, POLICY_PATH, scanners=["uc"]))
```

```python
# Step 3: scan. The summary shows scores, grade, counts, and anything it couldn't see.
result = run(POLICY_PATH, scanners=["uc"], spark=spark)
print(result.summary())
```

```python
# Step 4: one row per finding.
display(result.to_dataframe(spark))
```

```python
# Step 5 (optional): save the results. Leave both empty to only display them above.
SAVE_TABLE = ""  # e.g. "main.governance.lakehouse_gc_findings" (each run appends)
SAVE_JSON = ""  # e.g. f"/Volumes/main/default/lakehouse_gc/{result.run_id}.json" (new file)

if SAVE_TABLE or SAVE_JSON:
    for line in result.save(spark, table=SAVE_TABLE or None, json_path=SAVE_JSON or None):
        print("Saved", line)
else:
    print("Not saved: set SAVE_TABLE and/or SAVE_JSON to keep these results.")
```
